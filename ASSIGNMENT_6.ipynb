```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("a. Corrected temperatures:", corrected_temperature)

# b. Convert Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32
print("b. Fahrenheit:", fahrenheit)

# c. Identify readings greater than 32°C
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("c. Readings greater than 32°C:", greater_than_32)

# d. Count how many readings exceed 32°C
count = np.sum(corrected_temperature > 32)
print("d. Number of readings exceeding 32°C:", count)

# e. Explanation
print("""
e. Vectorization performs operations on the entire NumPy array at once,
without explicitly using a for loop. Boolean indexing allows us to
select elements satisfying a condition. These operations are faster,
simpler, and more efficient for processing large datasets.
""")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Calculate total steps
print("a. Total steps:", np.sum(steps))

# b. Calculate mean number of steps
print("b. Mean steps:", np.mean(steps))

# c. Maximum and minimum values
print("c. Maximum:", np.max(steps))
print("   Minimum:", np.min(steps))

# d. Total steps for each day (axis=0)
print("d. Total steps for each day:", np.sum(steps, axis=0))

# e. Total steps for each user (axis=1)
print("e. Total steps for each user:", np.sum(steps, axis=1))

# f. Position containing maximum number of steps
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("f. Maximum value:", np.max(steps))
print("   Position (row, column):", max_position)
```

```python
import numpy as np

# a. Create original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original:", original)

# b. Slice index 1 to 4
subset = original[1:5]
print("b. Subset:", subset)

# c. Modify first element of subset
subset[0] = 999

print("c. Original after modifying subset:", original)
print("   Subset:", subset)

# d. Create another slice using copy()
copied = original[1:5].copy()

copied[0] = 500

print("d. Original after modifying copied array:", original)
print("   Copied array:", copied)

# e. Create numbers 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)

print("e. 3 x 4 Matrix:")
print(matrix)

# f. Indexing and slicing

# First row
print("f. First row:", matrix[0])

# Last row
print("   Last row:", matrix[-1])

# Second column
print("   Second column:", matrix[:, 1])

# Rows 1-2 and columns 2-3
print("   Rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

# g. Flatten using flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()

print("g. Flatten:", flattened)
print("   Ravel:", raveled)

# h. Modify ravel()
raveled[0] = 100

print("h. Matrix after modifying ravel:")
print(matrix)
print("   Raveled array:", raveled)

# i. Modify flatten()
flattened[1] = 200

print("i. Matrix after modifying flatten:")
print(matrix)
print("   Flattened array:", flattened)

# j. Shape, dimensions, size and datatype
print("j. Shape:", matrix.shape)
print("   Dimensions:", matrix.ndim)
print("   Size:", matrix.size)
print("   Data type:", matrix.dtype)
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print("a. Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)

# b. Transpose
print("b. X transpose:")
print(X.T)

# c. Matrix product X.T @ X
XT_X = X.T @ X

print("c. X.T @ X:")
print(XT_X)

# d. Matrix inverse
inverse = np.linalg.inv(XT_X)

print("d. Inverse of X.T @ X:")
print(inverse)

# e. Ordinary Least Squares
beta = inverse @ X.T @ y

print("e. Coefficients (beta):")
print(beta)

# f. Explanation
print("""
f. The coefficients represent the contribution of each feature
to the predicted assistance score.

Beta[0] -> Sleep hours coefficient
Beta[1] -> Activity level coefficient
Beta[2] -> Stress level coefficient
""")

# g. Predict assistance score for a new user
new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print("g. Predicted assistance score:", prediction)
```

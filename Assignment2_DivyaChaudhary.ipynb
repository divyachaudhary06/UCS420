## Question 3 — Random Numbers with Lists

```python
# i.
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:", numbers)
```

```python
# ii.
odd_numbers = [n for n in numbers if n % 2 != 0]
print("Count of odd numbers:", len(odd_numbers))
```

```python
# iii.
even_numbers = [n for n in numbers if n % 2 == 0]
print("Count of even numbers:", len(even_numbers))
```

```python
# iv.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [n for n in numbers if is_prime(n)]
print("Count of prime numbers:", len(prime_numbers))
print("Prime numbers found:", prime_numbers)
```

```python
# v.
most_common = max(set(numbers), key=numbers.count)
print(f"Most frequent number: {most_common}, occurs {numbers.count(most_common)} time(s)")
```

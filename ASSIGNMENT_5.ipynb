```python
#b)
x=np.array([1,2,3,4,5,1,2,1,1,1])
y=np.array([1,1,1,2,3,4,2,4,3,3])

values, counts = np.unique(x, return_counts=True)
most_freq = values[np.argmax(counts)]

indices = np.where(x == most_freq)[0]

print("Most frequent value for x:", most_freq)
print("Indices for x:", indices)

values, counts = np.unique(y, return_counts=True)
most_freq = values[np.argmax(counts)]

indices = np.where(y == most_freq)[0]

print("\nMost frequent value for y:", most_freq)
print("Indices for y:", indices)
```

```python
UCS420_divya=np.array([10,20,30,40,
                      50,60,70,80,
                      90,15,20,35])
UCS420_divya.reshape(3,4)
print("mean:",np.mean(UCS420_divya))
print("median:",np.median(UCS420_divya))
print("min:",np.min(UCS420_divya))
print("max:",np.max(UCS420_divya))
print("unique:",np.unique(UCS420_divya))
```

```python
reshaped_ucs420_divya=UCS420_divya.reshape(4,3)
resized_ucs420_divya=np.resize(UCS420_divya,(2,3))
print(reshaped_ucs420_divya)
print(resized_ucs420_divya)
```
